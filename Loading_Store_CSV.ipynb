```python
# Load CSV file from volume and create Store_Sale table in learning schema

# Define environment parameter
dbutils.widgets.dropdown("environment", "dev", ["dev", "staging", "prod"], "Environment")
environment = dbutils.widgets.get("environment")

# Define csv file path parameter
dbutils.widgets.text("csv_file_path", "/Volumes/workspace/dev/csv", "CSV File Path")
csv_file_path = dbutils.widgets.get("csv_file_path")

# Read CSV file from volume
df = spark.read.csv(
    csv_file_path,
    header=True,
    inferSchema=True
)

# Clean column names - replace spaces with underscores and remove special characters
for col_name in df.columns:
    clean_name = col_name.strip().replace(" ", "_").replace("/", "_")
    df = df.withColumnRenamed(col_name, clean_name)

# Create the table in learning schema
df.write.mode("overwrite").saveAsTable(f"{environment}.Store_Sale")

# Display the table to verify
display(spark.table(f"{environment}.Store_Sale"))
#done
```
